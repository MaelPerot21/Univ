```python
#Selection sort

def selectionSort(listToSort) :
    i = 0
    while(i < len(listToSort)) :
        mini = i
        for k in range (i + 1, len(listToSort)) :
            if listToSort[k] < listToSort[mini] :
                mini = k
        save = listToSort[i]
        listToSort[i] = listToSort[mini]
        listToSort[mini] = save
        i += 1
    return listToSort
```

```python
#Insertion Sort

def insertionSort(listToSort) :
    i = 1
    while(i < len(listToSort)) :
        for k in range (i-1, -1, -1) :
            if listToSort[k] > listToSort[k+1] :
                save = listToSort[k+1]
                listToSort[k+1] = listToSort[k]
                listToSort[k] = save
        i += 1
    return listToSort 
```

```python
#Quick sort

def quickSort(listToSort) :
    if len(listToSort) > 1 :
        minusList = []
        majorList = []
        
        for i in range (1, len(listToSort)) :
            if listToSort[i] < listToSort[0] :
                minusList.append(listToSort[i])
            else :
                majorList.append(listToSort[i])
                
        return quickSort(minusList) + [listToSort[0]] + quickSort(majorList)
    
    else :
        return listToSort
```

```python
toSort = [3,4,12]
print(selectionSort(toSort))
print(insertionSort(toSort))
print(quickSort(toSort))
```
