```python
def file_fusion(*files) :
    result_file = open("melted_files", "a")
    check_result = open("melted_files", "r")
    
    for file in files :
        f = open(file, "r")
        line = f.readline()
        while line :
            if line not in check_result.read() :
                result_file.write(line)
            line = f.readline()
file_fusion("log_file.txt", "clients.json", "new_file" )
```

```python
import os

current_contain = os.listdir()
files_contained = []
for elem in current_contain :
    if os.path.isfile(elem) :
        files_contained += [elem]

for file in files_contained :
    i = -1
    #get the index of the last dot in variable i
    for k in range (0, len(file)) :
        if file[k] == "." :
            i = k
    
    #create the extension directory and move the file       
    if i == -1 or i == len(line)-1 :
        #no extension case
        if not os.path.exists("no_extension") :
            os.makedirs("no_extension")
        os.rename(file, "no_extension/"+file)
        
    else :
        ext = file[i+1:]
        if not os.path.exists(ext) :
            os.makedirs(ext)
        os.rename(file, ext+"/"+file)
```

```python
import os

def sort_files(path) :
    current_contain = os.listdir(os.getcwd()+path)
        
    files_contained = []
    for elem in current_contain :
        if os.path.isfile(os.getcwd()+((path+"/") if path else "/")+elem) :
            files_contained += [elem]
        elif os.path.isdir(os.getcwd()+((path+"/") if path else "/")+elem) :
            sort_files(path+"/"+elem)

    for file in files_contained :
        i = -1
        #get the index of the last dot in variable i
        for k in range (0, len(file)) :
            if file[k] == "." :
                i = k

        #create the extension directory and move the file       
        if i == -1 or i == len(line)-1 :
            #no extension case
            if not os.path.exists(os.getcwd()+((path+"/") if path else "/")+"no_extension") :
                os.makedirs(os.getcwd()+((path+"/") if path else "/")+"no_extension")
            os.rename(file, os.getcwd()+((path+"/") if path else "/")+"no_extension/"+file)

        else :
            ext = file[i+1:]
            if not os.path.exists(os.getcwd()+((path+"/") if path else "/")+ext) :
                os.makedirs(os.getcwd()+((path+"/") if path else "/")+ext)
            os.rename(file, os.getcwd()+((path+"/") if path else "/")+ext+"/"+file)
            
sort_files("")
```

```python
import os

def sort_files(path) :
    current_contain = os.listdir(os.getcwd()+path)
        
    files_contained = []
    for elem in current_contain :
        if os.path.isfile(os.getcwd()+((path+"/") if path else "/")+elem) :
            if not elem in files_contained : 
                files_contained += [elem]
        elif os.path.isdir(os.getcwd()+((path+"/") if path else "/")+elem) :
            sort_files(path+"/"+elem)

    for file in files_contained :
        i = -1
        #get the index of the last dot in variable i
        for k in range (0, len(file)) :
            if file[k] == "." :
                i = k

        #create the extension directory and move the file       
        if i == -1 or i == len(line)-1 :
            #no extension case
            if not os.path.exists(os.getcwd()+((path+"/") if path else "/")+"no_extension") :
                os.makedirs(os.getcwd()+((path+"/") if path else "/")+"no_extension")
            os.rename(file, os.getcwd()+((path+"/") if path else "/")+"no_extension/"+file)

        else :
            ext = file[i+1:]
            if not os.path.exists(os.getcwd()+((path+"/") if path else "/")+ext) :
                os.makedirs(os.getcwd()+((path+"/") if path else "/")+ext)
            os.rename(file, os.getcwd()+((path+"/") if path else "/")+ext+"/"+file)
            
sort_files("")
```
