```python
n = int(input("Combien de termes de la suite de Fibonacci voulez-vous : "))

minus2Term = 0
minus1Term = 1
for i in range(0, n+1) :
    if i == 0 :
        print(0)
    elif i == 1 :
        print(1)
    else :
        nTerm = minus1Term + minus2Term
        minus2Term = minus1Term
        minus1Term = nTerm
        print(nTerm)
    
```

```python
num = int(input("De quel nombre faut-il la racine ? : "))
epsilon = int(input("Avec quel indice de precision la determiner ? : "))

k = 0
#This loop finds the integer part of num's square root
while k*k <= num :
    k +=1

while epsilon > 0 :
    #recurrence given by the Newton's method 
    k = (k + num/k) / 2
    epsilon -=1

print(str(k) + " est la racine de " + str(num))
```

```python
userList = []
lastNumber = "n"
while lastNumber != "y" :
    number = int(input("Ajoutez un nombre a la liste ('exit' pour arreter) : "))
    userList += [number]
    
    lastNumber = input("Etait-ce le dernier nombre ? ('y' pour oui) : ")

changed = 1
while changed == 1 :
    changed = 0
    for k in range (0, len(userList) - 1) :
        if userList[k] > userList[k+1] :
            save = userList[k]
            userList[k] = userList[k+1]
            userList[k+1] = save
            changed = 1
    
print(userList)
```

```python
amount = int(input("Quel est le montant a retirer ? : "))
money = [20, 10, 5, 1]
nbToGive = 0

if amount < 0 :
    print("Le montant ne peut etre distribue")
else :
    moneyInd = 0
    while amount > 0 :
        while amount >= money[moneyInd] :
            amount -=money[moneyInd]
            nbToGive +=1
        moneyInd +=1
        
print(nbToGive)
```
