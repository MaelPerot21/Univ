```python
num = int(input("Saisir un nombre : "))

if num > 0 :
    print("Le nombre est positif")
elif num == 0 :
    print("Le nombre est nul")
else :
    print("Le nombre est negatif")
```

```python
firstNum = int(input("Entrez un nombre : "))
secondNum = int(input("Entrez un nombre : "))
thirdNum = int(input("Entrez un nombre : "))

if firstNum > secondNum :
    if firstNum > thirdNum :
        print(str(firstNum) + " est le plus grand")
    else :
        print(str(thirdNum) + " est le plus grand")

elif secondNum > thirdNum :
    print(str(secondNum) + " est le plus grand")
else :
    print(str(thirdNum) + " est le plus grand")
```

```python
n = int(input("Combien de termes de la suite de Fibonacci voulez-vous : "))

minus2Term = 0
minus1Term = 1
for i in range(0, n+1) :
    if i == 0 :
        print(0)
    elif i == 1 :
        print(1)
    else :
        nTerm = minus1Term + minus2Term
        minus2Term = minus1Term
        minus1Term = nTerm
        print(nTerm)
    
```

```python
import random 

toGuess = random.randint(0, 100)
guess = -1

while guess != toGuess :
    guess = int(input("Quel est le nombre auquel je pense ? : "))
    if guess > toGuess :
        print("Trop grand")
    elif guess < toGuess :
        print("Trop petit")
    else :
        print("Correct")
```
