```python
class Livre :
    def __init__(self, titre, auteur, nombre_pages) :
        self.titre = titre
        self.auteur = auteur
        self.nombre_pages = nombre_pages
    def afficher(self) :
        print(self.titre + " est ecrit par " + self.auteur + " et fait " + str(self.nombre_pages) + " pages")
    def compare_page_amount(self, livre) :
        if self.nombre_pages == livre.nombre_pages :
            print("Le nombre de pages est identique")
        elif self.nombre_pages > livre.nombre_pages :
            print(self.titre + " a plus de pages")
        else :
            print(livre.titre + " a plus de pages")
            
livre1 = Livre("titre1", "auteur1", 30)
livre2 = Livre("titre2", "auteur2", 40)

livre1.compare_page_amount(livre2)
```

```python
class Livre :
    def __init__(self, titre, auteur, nombre_pages) :
        self.titre = titre
        self.auteur = auteur
        self.__setNombrePages(nombre_pages)
        
    def getNombrePages(self) :
        return self._nombre_pages
    def __setNombrePages(self, value) :
        if value > 0 and not hasattr(self, '_nombre_pages'):
            self._nombre_pages = value
        
    def afficher(self) :
        print(self.titre + " est ecrit par " + self.auteur + " et fait " + str(self._nombre_pages) + " pages")
    def compare_page_amount(self, livre) :
        if self._nombre_pages == livre.getNombrePages() :
            print("Le nombre de pages est identique")
        elif self._nombre_pages > livre.getNombrePages() :
            print(self.titre + " a plus de pages")
        else :
            print(livre.titre + " a plus de pages")
            
livre1 = Livre("titre1", "auteur1", 30)
livre2 = Livre("titre2", "auteur2", 40)

livre1.compare_page_amount(livre2)
```

```python
class Roman(Livre) :
    def __init__(self, titre, auteur, nombre_pages, genre) :
        super().__init__(titre, auteur, nombre_pages)
        self.genre = genre
    def afficher(self) :
        print("genre : " + self.genre, end = ", ")
        super().afficher()
        
test2 = Roman("Les miserables", "Victor Hugo", 402, "realisme")
test2.afficher()
```

```python
class Magazine(Livre) :
    def __init__(self, titre, auteur, nombre_pages, periodicite) :
        super().__init__(titre, auteur, nombre_pages)
        self.periodicite = periodicite
    def afficher(self) :
        print("De facon " + self.periodicite, end = ", ")
        super().afficher()        
```

```python
class Livre :
    def __init__(self, titre, auteur, nombre_pages) :
        self.titre = titre
        self.auteur = auteur
        self.__setNombrePages(nombre_pages)
        
    def getNombrePages(self) :
        return self._nombre_pages
    def __setNombrePages(self, value) :
        if value > 0 and not hasattr(self, '_nombre_pages'):
            self._nombre_pages = value
            
    def __str__(self) :
        return "titre:" + self.titre + " ; auteur:" + self.auteur + " ; nombre de pages:" + str(self._nombre_pages)
      
    def __add__(self, livre) :
        return Livre(self.titre + livre.titre, self.auteur + livre.auteur, self._nombre_pages + livre._nombre_pages)
        
    def afficher(self) :
        print(self.titre + " est ecrit par " + self.auteur + " et fait " + str(self._nombre_pages) + " pages")
   
    def compare_page_amount(self, livre) :
        if self._nombre_pages == livre.getNombrePages() :
            print("Le nombre de pages est identique")
        elif self._nombre_pages > livre.getNombrePages() :
            print(self.titre + " a plus de pages")
        else :
            print(livre.titre + " a plus de pages")

livre1 = Livre("titre1", "auteur1", 30)
print(livre1)
livre2 = Livre("titre2", "auteur2", 40)
print(livre1 + livre2)
```
