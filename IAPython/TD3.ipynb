```python
def est_pair(n) :
    if n % 2 == 0 :
        return True
    else :
        return False
    
est_pair(1)
```

```python
def convertir_temperature(t, unite) :
    if unite == "C" :
        return t * (9/5) + 32
    elif unite == "F" :
        return (t - 32) / (9/5)
    else :
        print("L'unite n'est pas valide, il faut entrer 'C' ou 'F'")
    
convertir_temperature(50, "F")
```

```python
def racine(n, precision = 0.0001) :
    res = 0
    while (res+1)*(res+1) <= n :
        res +=1
    while (res+precision) * (res+precision) <= n :
        res +=precision
    
    return res

racine(2)
```

```python
def maximum(a, b, c) :
    if a > b :
        if a > c :
            return a
        else :
            return c
    elif c > b :
            return c
    else :
            return b
        
maximum(1, 3, 2)
```

```python
def moyenne(liste_nombre) :
    total = 0
    for i in range (0, len(liste_nombre)) :
        total += liste_nombre[i]
    return total / len(liste_nombre)

moyenne([1,2,3,4,5])
```

```python
def calculer_salaire(heures, taux_horaire, prime=0) :
    return heures * taux_horaire + prime

calculer_salaire(35, 12, 50)
```

```python
def pgcd(a, b) :
    if b == 0 :
        return a
    
    reste = a % b
    a = b
    b = reste
    
    return pgcd(a, b)

pgcd(12, 9)
```

```python
def somme(*nombres) :
    res = 0
    for i in range (0, len(nombres)) :
        res += nombres[i]
    
    return res

somme(1,2,3,4,5)
```

```python
def concatener(*mots) :
    if len(mots) > 0 :
        sentence = mots[0]
    for i in range (1, len(mots)) :
        sentence += " " + mots[i]
        
    return sentence

concatener("Phrase", "avec", "des", "espaces.")
```

```python
def statistiques(*nombres) :
    if len(nombres) > 0 :
        maxi = nombres[0]
        somme = maxi
        for i in range(1, len(nombres)) :
            somme += nombres[i]
            if nombres[i] > maxi :
                maxi = nombres[i]
        return somme, somme / len(nombres), maxi
    else :
        print("Les statistiques ne peuvent pas etre caclulees sans arguments")
        
statistiques(1,2,3,4,5)
```

```python
def decrire_personne(nom, age, ville="Inconnue") :
    if (ville != "Inconnue") :
        print("Monsieur " + nom + " a " + str(age) + " ans et habite a " + ville)
    else :
        print("Monsieur " + nom + " a " + str(age) + " ans")
```

```python
decrire_personne("Dupont", 55)
decrire_personne("Dupont", 55, "Paris")
```

```python
def decrire_personne_v2(**kwargs) :
    print(kwargs)
    
decrire_personne_v2(fName="Jean", lName="Dupont", age=55, town="Paris")
```

```python
def fusionner_dicts(**dicts) :
    resDict = {}
    for dict, dictVar in dicts.items() :
        for key, value in dictVar.items() :
            if key in resDict :
                resDict[key] += [value]
            else :
                resDict[key] = [value]
    return resDict

dict1 = {'fName': 'Jean', 'lName': 'Dupont', 'age': 55, 'town': 'Paris'}
dict2 = {'fName': 'John', 'lName': 'Bridge', 'age': 99, 'town': 'London'}
fusionner_dicts(fDict = dict1, sDict = dict2)
```
